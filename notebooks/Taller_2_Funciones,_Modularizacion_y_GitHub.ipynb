## Parte 1 — Función para convertir valores

Cree una función llamada `convertir_a_float(valor)` que intente convertir un valor a número decimal.

La función debe retornar:

- El número convertido, si la conversión es posible.
- `None`, si el valor no se puede convertir.

````python
# TODO: cree la función convertir_a_float

def convertir_a_float(valor):
    try:
        return float(valor)
    except (ValueError, TypeError):
        return None

# Pruebas sugeridas
print(convertir_a_float("12.5"))
print(convertir_a_float("error"))
print(convertir_a_float(""))
print(convertir_a_float(None))
````

## Parte 2 — Función para calcular potencia

Cree una función llamada `calcular_potencia(voltaje, corriente)`.

La función debe:

- Retornar `None` si voltaje o corriente son `None`.
- Retornar `None` si voltaje o corriente son negativos.
- Retornar `voltaje * corriente` en los demás casos.

````python
# TODO: cree la función calcular_potencia

def calcular_potencia(voltaje, corriente):
    if voltaje is None or corriente is None:
        return None
    if voltaje < 0 or corriente < 0:
        return None
    return voltaje * corriente

# Pruebas sugeridas
print(calcular_potencia(12, 2))
print(calcular_potencia(None, 2))
print(calcular_potencia(12, -1))
````

## Parte 3 — Función para clasificar temperatura

Cree una función llamada `clasificar_temperatura(temp)` con los siguientes criterios:

- Si `temp` es `None`: `"Dato inválido"`
- Si `temp < 40`: `"Normal"`
- Si `40 <= temp < 50`: `"Precaución"`
- Si `temp >= 50`: `"Alerta"`

````python
# TODO: cree la función clasificar_temperatura

def clasificar_temperatura(temp):
    if temp is None:
        return "Dato inválido"
    elif temp < 40:
        return "Normal"
    elif temp < 50:
        return "Precaución"
    else:
        return "Alerta"

# Pruebas sugeridas
print(clasificar_temperatura(35))
print(clasificar_temperatura(45))
print(clasificar_temperatura(52))
print(clasificar_temperatura(None))
````

## Parte 4 — Limpieza y procesamiento de registros

Procese la lista `registros_crudos` y cree una nueva lista llamada `registros_limpios`.

Cada registro limpio debe contener:

- `id`
- `voltaje`
- `corriente`
- `temperatura`
- `potencia`
- `estado_temperatura`

Use las funciones creadas.

````python
registros_limpios = []

for registro in registros_crudos:
    voltaje = convertir_a_float(registro["voltaje"])
    corriente = convertir_a_float(registro["corriente"])
    temperatura = convertir_a_float(registro["temperatura"])

    registros_limpios.append({
        "id": registro["id"],
        "voltaje": voltaje,
        "corriente": corriente,
        "temperatura": temperatura,
        "potencia": calcular_potencia(voltaje, corriente),
        "estado_temperatura": clasificar_temperatura(temperatura)
    })

registros_limpios
````

## Parte 5 — Comprensión de listas

Use comprensión de listas para obtener:

````python
# TODO: use comprensión de listas

potencias_validas = [
    registro["potencia"]
    for registro in registros_limpios
    if registro["potencia"] is not None
]

temperaturas_validas = [
    registro["temperatura"]
    for registro in registros_limpios
    if registro["temperatura"] is not None
]

motores_alerta = [
    registro
    for registro in registros_limpios
    if registro["estado_temperatura"] == "Alerta"
]

print(potencias_validas)
print(temperaturas_validas)
print(motores_alerta)
````

## Parte 7 — Resumen automático

Cree una función llamada `generar_resumen(registros)` que retorne un diccionario con:

- `total_registros`
- `registros_validos_potencia`
- `potencia_promedio`
- `temperatura_promedio`
- `cantidad_alertas`
- `cantidad_precauciones`
- `cantidad_normales`

````python
# TODO: cree la función generar_resumen

def generar_resumen(registros):
    potencias = [
        registro["potencia"]
        for registro in registros
        if registro["potencia"] is not None
    ]

    temperaturas = [
        registro["temperatura"]
        for registro in registros
        if registro["temperatura"] is not None
    ]

    return {
        "total_registros": len(registros),
        "registros_validos_potencia": len(potencias),
        "potencia_promedio": sum(potencias) / len(potencias) if potencias else None,
        "temperatura_promedio": sum(temperaturas) / len(temperaturas) if temperaturas else None,
        "cantidad_alertas": sum(1 for registro in registros if registro["estado_temperatura"] == "Alerta"),
        "cantidad_precauciones": sum(1 for registro in registros if registro["estado_temperatura"] == "Precaución"),
        "cantidad_normales": sum(1 for registro in registros if registro["estado_temperatura"] == "Normal")
    }

resumen = generar_resumen(registros_limpios)
resumen
````
