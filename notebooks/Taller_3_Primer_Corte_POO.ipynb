## Punto 1 – Diseño orientado a objetos y abstracción (1.0)

Diseñe una clase abstracta `Sensor` que represente un sensor genérico.
- Atributo protegido `_id_sensor`
- Atributo protegido `_unidad`
- Método abstracto `leer()` que retorne una medición
- Método concreto `info()` que retorne una cadena con el ID y la unidad

Implemente dos clases concretas `SensorTemperatura` y `SensorPresion`, cada una sobrescribiendo `leer()` con un valor simulado.

```python
import random
from abc import ABC, abstractmethod

class Sensor(ABC): #clase abstracta(clase plantilla de otras clases)
  def __init__(self, id_sensor, unidad):
    self._id_sensor=id_sensor
    self._unidad=unidad

  @abstractmethod
  def leer(self):
    pass

  def info(self):
    return self._id_sensor +" "+ self._unidad

class SensorTemperatura(Sensor):
  def __init__(self, id_sensor):
    super().__init__(id_sensor, "C")

  def leer(self):
    return random.uniform(10,30)

class SensorPresion(Sensor):
  def __init__(self, id_sensor):
    super().__init__(id_sensor, "kPa")

  def leer(self):
    return random.uniform(30,50)

st1=SensorTemperatura("TEMP-01")
sp1=SensorPresion("PRES-01")

print(st1.info())
print(st1.leer())
print(sp1.info())
print(sp1.leer())
```

## Punto 2 – Encapsulamiento y validación con propiedades (1.0)

Cree una clase `CuentaBancaria` con:
- Atributo privado `__saldo`
- Métodos `depositar(monto)` y `retirar(monto)`
- Propiedad `saldo` de solo lectura

Condiciones:
- No se permite retirar más dinero del saldo disponible
- No se permiten montos negativos

```python
class CuentaBancaria:
  def __init__(self, saldo_inicial=0):
    self._saldo=saldo_inicial

  def depositar(self,monto):
    if monto>0:
      self._saldo += monto
    else:
      print("No se permiten montos negativos")

  def retirar(self,monto):
     if monto<=0:
      print("No se permiten montos negativos")
     elif monto> self._saldo:
      print("Saldo insuficiene")
     else:
      self._saldo -= monto

  @property
  def saldo(self):
    return self._saldo

cuenta= CuentaBancaria(1000)
print(cuenta.saldo)

cuenta.depositar(500)
print(cuenta.saldo)

cuenta.retirar(300)
print(cuenta.saldo)

cuenta.retirar(1500)
cuenta.depositar(-100)
```

## Punto 3 – Polimorfismo y listas de objetos (1.0)

Dada una lista de objetos de tipo `Sensor` (temperatura y presión), implemente una función `leer_sensores(lista_sensores)` que:
- Recorra la lista
- Llame al método `leer()` de cada objeto
- Imprima la información del sensor y su lectura

```python
def leer_sensores(lista_sensores):
  for sensor in lista_sensores:
    print(sensor.info())
    print("Lectura:",sensor.leer())

st1 = SensorTemperatura("TEMP-01")
st2 = SensorTemperatura("TEMP-02")

sp1 = SensorPresion("PRES-01")
sp2 = SensorPresion("PRES-02")

lista_sensores=[st1,st2,sp1,sp2]
leer_sensores(lista_sensores)
```

## Punto 4 – Composición vs herencia (1.0)
### Enunciado
Implemente una clase `SistemaMonitoreo` que:
- Contenga una lista de sensores (composición)
- Tenga un método `agregar_sensor(sensor)`
- Tenga un método `reporte()` que muestre todas las lecturas

Justifique brevemente por qué aquí es mejor composición que herencia.

```python
class SistemaMonitoreo:
  def __init__(self):
   self.sensores=[]

  def agregar_sensor(self,sensor):
   self.sensores.append(sensor)

  def reporte(self):
   for sensor in self.sensores:
    print(sensor.info())
    print("lectura:",sensor.leer())

sistema = SistemaMonitoreo()

st1 = SensorTemperatura("TEMP-01")
sp1 = SensorPresion("PRES-01")

sistema.agregar_sensor(st1)
sistema.agregar_sensor(sp1)

sistema.reporte()
```

Se usa composición porque un SistemaMonitoreo tiene varios sensores, pero no es un tipo de sensor. La herencia se usaría si SistemaMonitoreo fuera una clase especializada de Sensor, lo cual no tendría sentido

## Punto 5 – Gestión de estados y responsabilidad única (SRP) (1.0)

Diseñe una clase `Experimento` que represente un experimento computacional o de laboratorio.

Requisitos:
1. Atributos: `nombre` y `estado` (solo: `'creado'`, `'configurado'`, `'ejecutado'`).
2. Métodos:
   - `configurar()`: solo si el estado es `'creado'`
   - `ejecutar()`: solo si el estado es `'configurado'`
   - `reiniciar()`: vuelve a `'creado'`
3. Si una acción no está permitida, lanzar una excepción con mensaje claro.
4. Método `info()` que muestre nombre y estado.

```python
class Experimento:
  def __init__(self,nombre, estado):
   self.nombre=nombre
   self.estado=estado

  def configurar(self):
    if self.estado=="creado":
     self.estado="configurado"
    else:
      raise Exception("No se puede configurar porque el experimento no esta creado")

  def ejecutar(self):
    if self.estado=="configurado":
      self.estado="ejecutado"
    else:
     raise Exception("No se puede ejecutar porque no esta configutado")

  def reiniciar(self):
    if self.estado=="ejecutado":
      self.estado="creado"

  def info(self):
    print("nombre:", self.nombre," ","estado:",self.estado)

exp1=Experimento("Prueba temperatura","creado")
exp1.info()
exp1.configurar()
exp1.info()
exp1.ejecutar()
exp1.info()
exp1.reiniciar()
exp1.info()
```
