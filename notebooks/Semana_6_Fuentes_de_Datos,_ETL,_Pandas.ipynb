## 9. Función para diagnóstico inicial del dataset

Cuando se carga un dataset, conviene aplicar una revisión inicial estándar.

````python
def diagnostico_inicial(df, nombre_dataset="dataset"):
    print(f"Diagnóstico inicial: {nombre_dataset}")
    print("-" * 50)
    print("Dimensiones:", df.shape)
    print("\nColumnas:")
    print(df.columns.tolist())
    print("\nTipos de datos:")
    display(df.dtypes)
    print("\nPrimeras filas:")
    display(df.head())
    print("\nValores faltantes:")
    display(df.isna().sum())
    print("\nEstadísticas descriptivas:")
    display(df.describe(include="all"))

diagnostico_inicial(df_archivo_propio, "df_archivo_propio")
````

## 10. Transformación mínima

En un flujo ETL inicial no siempre se hace limpieza profunda.  
La transformación mínima puede incluir:

- Normalizar nombres de columnas.
- Agregar una columna de fuente.
- Calcular una variable simple.
- Seleccionar columnas relevantes.
- Guardar una versión preliminar.

````python
def normalizar_columnas(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )
    return df

df_etl = normalizar_columnas(df_archivo_propio)
df_etl["fuente"] = "archivo_propio"
df_etl["potencia_W"] = df_etl["voltaje_v"] * df_etl["corriente_a"]

df_etl
````

## 13. Flujo ETL completo en una función

Podemos encapsular el flujo básico en una función.

````python
def etl_archivo_csv(ruta_entrada, ruta_salida, fuente):
    # Extract
    df = pd.read_csv(ruta_entrada)

    # Transform
    df = normalizar_columnas(df)
    df["fuente"] = fuente

    if "voltaje_v" in df.columns and "corriente_a" in df.columns:
        df["potencia_w"] = df["voltaje_v"] * df["corriente_a"]

    # Load
    df.to_csv(ruta_salida, index=False)

    return df

df_resultado = etl_archivo_csv(
    ruta_entrada="data/raw/mediciones_propias.csv",
    ruta_salida="data/processed/mediciones_propias_etl_funcion.csv",
    fuente="archivo_propio"
)

df_resultado
````
